# Проверка цены WB

Используется метод цен `/__internal/card/cards/v4/detail`. Введите данные существующей браузерной сессии локально. Этот ноутбук не получает и не обновляет токен автоматически. Код не отправляет `Authorization` или cookies кабинета.


```python
from pathlib import Path
import sys

repo_root = next((path for path in (Path.cwd(), Path.cwd().parent) if (path / "wb_prices.py").is_file()), None)
if repo_root is None:
    raise RuntimeError("Откройте ноутбук из папки great_equalizer или great_equalizer/notebooks")
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from getpass import getpass
from wb_prices import BrowserContext, fetch_prices
```

Введите токен и `deviceid` через скрытый ввод. User-Agent должен соответствовать источнику токена. Для сравнения цен сохраняйте один и тот же `dest`. Не вставляйте реальные значения в исходный код ячеек.


```python
token = getpass("x_wbaas_token: ")
deviceid = getpass("deviceid: ")
user_agent = input("User-Agent из того же запроса: ").strip()
dest = input("dest [-1257786]: ").strip() or "-1257786"
context = BrowserContext.create(token, deviceid, user_agent)
del token, deviceid
```

```python
articles = [732445767, 732449137, 386287758, 1247697027, 1050925338, 804787539]
result = fetch_prices(articles, context, dest=dest)
result
```

`price_after_spp_rub` — цена из ответа WB. `wallet_price_estimate_rub` — расчёт −3% с округлением до рубля, не готовое поле WB. `complete=False` означает, что часть данных отсутствует. HTTP 403/498 не доказывает необходимость входа: эти ошибки требуют отдельной проверки сессии.
